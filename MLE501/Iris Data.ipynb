```python
from sklearn.datasets import load_iris
import pandas as pd
import numpy as np

iris = load_iris()

df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['name'] = iris.target_names[iris.target]
df.head()
```

```python
sepal_lengths=df.iloc[:,0]
print(sepal_lengths)

min_sepal_length=sepal_lengths.min()
max_sepal_length=sepal_lengths.max()

# min max scaling
min_max_scaling = (sepal_lengths - min_sepal_length) / (max_sepal_length - min_sepal_length)
print(min_max_scaling)

# mean normalization
mean_sepal_length=sepal_lengths.mean()
std_sepal_length=sepal_lengths.std()

mean_normalization = (sepal_lengths - mean_sepal_length) / std_sepal_length
print(mean_normalization)




```
