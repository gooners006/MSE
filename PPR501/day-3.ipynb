```python
tour_prices = {
    "Italy": 2500,
    "France": 3000,
    "Japan": 4000,
    "Egypt": 2800,
    "Costa Rica": 3500,
    "Peru": 2200,
    "Thailand": 2700,
    "Tanzania": 3800
}

display(tour_prices)
```

```python
prompt = input("Enter a country or price to find a tour: ")
if prompt.isnumeric():
    for country, price in tour_prices.items():
        if price == int(prompt):
            print(f"There is the tour to {country} for the price of {prompt}")
            break
        else:
            continue
    else:
        print(f"Sorry, there is no tour for the price of {prompt}")
elif prompt.isalpha():
    if prompt in tour_prices:
        print(f"There is the tour to {prompt} for the price of {tour_prices[prompt]}")
    else:
        print(f"Sorry, there is no tour to {prompt}")
```

```python
# Question 10 : Write a program to input an English sentence and translate it into Vietnamese
# The input will stop when the user input No
from deep_translator import GoogleTranslator
import sys

# Initialize the translator
target_prompt = input(
    "Select the target language (type `vi` for Vietnamese, `en` for English): "
)
while target_prompt not in ["vi", "en"]:
    target_prompt = input(
        "Invalid input. Select the target language (type `vi` for Vietnamese, `en` for English, `exit` to quit the program): "
    )
    if target_prompt.lower() == "exit":
        print("Thank you for using the translator!")
        sys.exit()


source_map = {
    "vi": {
        "source": "en",
        "target": "vi",
        "welcomeText": "Welcome to the English-to-Vietnamese Translator!",
    },
    "en": {
        "source": "vi",
        "target": "en",
        "welcomeText": "Welcome to the Vietnamese-to-English Translator!",
    },
}

translator = GoogleTranslator(
    source=source_map[target_prompt]["source"],
    target=source_map[target_prompt]["target"],
)

print(source_map[target_prompt]["welcomeText"])
print("Type 'exit' to stop the program.\n")
translate_prompt = input("Enter a sentence: ")
while translate_prompt.lower() != "exit":
    print(f"Translated: {translator.translate(translate_prompt)}\n")
    translate_prompt = input(
        "Enter another sentence (type `exit` to stop the program): "
    )
else:
    print("Thank you for using the translator!")
```
