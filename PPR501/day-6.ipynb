```python
class Student:
    number_of_students = 0
    def __init__(self, name, age, height):
        self.name = name
        self.age = age
        self.height = height
        Student.number_of_students += 1

    def info(self):
        return f"Name: {self.name}, Age: {self.age}, Height: {self.height} cm"

    def improveHeight(self, height):
        self.height += height


Van = Student("Van", 27, 168)
print(Van.height)
print(Van.info())
Van.improveHeight(5)
print(Van.info())
```

```python
%%writefile class_inheritance.py
class A:
    def top(self):
        print("top")


class B(A):
    def middle(self):
        print("middle")


class C(B):
    'Hello C'
    def bottom(self):
        print("bottom")

if __name__ == "__main__":
    obj=C()
    obj.top()
    obj.middle()
    obj.bottom()
```

```python
import class_inheritance

c = class_inheritance.C()
c.top()
c.middle()
c.bottom()
```

```python
from class_inheritance import C

c2 = C()
c2.top()
c2.middle()
c2.bottom()
```

```python
import os

content = """
# Một lớp mô phỏng một hình chữ nhật.
class Rectangle :
    'This is Rectangle class'
    # Một phương thức được sử dụng để tạo đối tượng (Contructor).
    def __init__(self, width, height):
        self.width= width
        self.height = height
    def getWidth(self):        
        return self.width
    def getHeight(self):        
        return self.height
    # Phương thức tính diện tích.
    def getArea(self):
        return self.width * self.height
"""

file = path_Lib + "/rectlib.py"
with open(file, "w", encoding="utf-8") as filetowrite:
    filetowrite.write(content)

checkFile = os.path.isfile(file)
print("The path and file are valid or not :", checkFile)
print(file)
```

```python
import rectlib

rect = rectlib.Rectangle(10, 25)
print(rect.getArea())
```
