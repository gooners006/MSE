```python
import cv2, numpy

cap = cv2.VideoCapture(0)
# cap = cv2.VideoCapture("video.mp4")

while True:
    ret, frame = cap.read()
    cv2.imshow("Original", frame)
    edges = cv2.Canny(frame, 100, 200, True)
    cv2.imshow("Edges", edges)
    if cv2.waitKey(1) == ord("q"):
        break

cap.release()
```
