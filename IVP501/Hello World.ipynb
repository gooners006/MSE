## Translation

```python
import numpy as np

tx = 3
ty = 2

T = np.array((
  (1, 0, tx),
  (0, 1, ty),
  (0, 0, 1)
  ))

print(T)
```

```python
A = np.array((2, 3, 1))
B = np.array((4, 3, 1))
C = np.array((2, 1, 1))

G=np.array([A,B,C,A])

print(G)
```

```python
plt.plot(G[:, 0], G[:, 1], marker="o")
plt.xlim(0, 8)
plt.ylim(0, 8)
plt.grid(True)
plt.show()
```

```python
print(G[0, :])
print(T.dot(G[0, :]))
print(G.shape)
```

```python
G_t = np.zeros_like(G)
for i in range(G.shape[0]):
    G_t[i] = T.dot(G[i, :])

print(G_t)

plt.plot(G[:, 0], G[:, 1], marker="o")
plt.plot(G_t[:, 0], G_t[:, 1], marker="o")
plt.xlim(0, 8)
plt.ylim(0, 8)
plt.grid(True)
plt.show()
```

## Scale

```python
tx = 0
ty = 0
sx = 2
sy = 1.5

S = np.array((
  (sx, 0, tx), 
  (0, sy, ty), 
  (0, 0, 1)
  ))

print(S)
```

```python
G_s=np.zeros_like(G)
for i in range(G.shape[0]):
    G_s[i]=S.dot(G[i,:])

print(G_s)
```

```python
plt.plot(G[:, 0], G[:, 1], marker="o")
plt.plot(G_t[:, 0], G_t[:, 1], marker="o")
plt.plot(G_s[:, 0], G_s[:, 1], marker="o")
plt.xlim(0, 10)
plt.ylim(0, 10)
plt.grid(True)
plt.show()
```

## Rotation

```python
G_r=np.zeros_like(G)
theta=np.radians(90)
R=np.array(((np.cos(theta),-np.sin(theta),0),
            (np.sin(theta),np.cos(theta),0),
            (0,0,1)))

print(R)
```

```python
for i in range(G.shape[0]):
    G_r[i]=R.dot(G[i,:])

print(G_r)
```

```python
plt.plot(G[:, 0], G[:, 1], marker="o")
plt.plot(G_t[:, 0], G_t[:, 1], marker="o")
plt.plot(G_s[:, 0], G_s[:, 1], marker="o")
plt.plot(G_r[:, 0], G_r[:, 1], marker="o")
plt.xlim(-4, 10)
plt.ylim(0, 10)
plt.grid(True)
plt.show()
```
