```python
import os
import cv2
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
train_anno_paths=[]
valid_anno_paths=[]
for dirname, _, filenames in os.walk('./'):
    for filename in filenames:
        if 'train' in dirname and '.txt' in filename:
            train_anno_paths+=[(os.path.join(dirname, filename))]
        if 'valid' in dirname and '.txt' in filename:
            valid_anno_paths+=[(os.path.join(dirname, filename))]

print(valid_anno_paths[0:1])
```

```python
for path in train_anno_paths[0:1]:
    img_path=path[0:-4]+'.jpg'
    with open(path, 'r') as file:
        text = file.read().strip().split("\n")
    data = [line.split() for line in text]
    box = pd.DataFrame(data)
    display(box)
    img=plt.imread(img_path)
    plt.imshow(img)
    plt.axis('off')
    plt.show()
```

```python
def draw_box(anno_path):
    img_path=anno_path[0:-4]+'.jpg'
    with open(anno_path, 'r') as file:
        text = file.read().strip().split("\n")
    data = [line.split() for line in text]
    box = pd.DataFrame(data)
    box = box.astype(float)
    image = plt.imread(img_path)
    image_copy = image.copy()
    H,W = image_copy.shape[0],image_copy.shape[1]
    #print(H,W)
    for i in range(len(box)):
        label=box.loc[i,0]
        xc=int(box.loc[i,1]*W)
        yc=int(box.loc[i,2]*H)
        w=int(box.loc[i,3]*W) 
        h=int(box.loc[i,4]*H)
        y1=yc-int(0.5*h)
        y2=yc+int(0.5*h)
        x1=xc-int(0.5*w)
        x2=xc+int(0.5*w)
        cv2.rectangle(image_copy,(x1,y1),(x2,y2),(0,255,0),5) #green
    
    plt.imshow(image_copy)
    plt.axis('off')
    plt.show()   
```

```python
paths=random.sample(train_anno_paths,20)
for path in paths:
    draw_box(path)
```
