## White Background

```python
import numpy as np
import cv2

white_bg = np.ones((1080, 1920, 3), dtype=np.uint8) * 255

# Create a named window and make it resizable
cv2.namedWindow("White Background", cv2.WINDOW_NORMAL)

while True:
    # Display the image
    cv2.imshow("White Background", white_bg)
    # Wait for a key press with a small delay (1ms)
    key = cv2.waitKey(1)
    # If 'q' or ESC is pressed, break the loop
    if key == ord("q") or key == 27:  # 27 is the ASCII code for ESC
        break

cv2.destroyAllWindows()
```
