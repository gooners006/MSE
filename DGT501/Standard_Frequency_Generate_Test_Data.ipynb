```python
import os
import numpy as np
import soundfile as sf
from scipy.signal import resample
import random

# Cấu hình thư mục
INPUT_FOLDER = "D:\\MSE23\\XLTHS\\Musical Instrument Tuning\\archive\\Guitar Dataset\\E2"
OUTPUT_FOLDER = "D:\\MSE23\\XLTHS\\Musical Instrument Tuning\\archive\\Guitar Dataset\\E2_Test_SF"
E2_FREQ = 82.41  # Tần số chuẩn của nốt E2

def cents_to_factor(cents):
    """Chuyển đổi sai số cents sang hệ số tần số."""
    return 2 ** (cents / 1200)

def adjust_pitch(signal, factor):
    """Điều chỉnh tần số của âm thanh dựa trên hệ số."""
    new_length = int(len(signal) / factor)
    return resample(signal, new_length)

def create_test_files(file_path):
    """Tạo file test từ file đầu vào, dựa trên tần số chuẩn E2."""
    # Đọc file âm thanh
    data, sample_rate = sf.read(file_path)
    if len(data.shape) > 1:  # Nếu là stereo
        data = np.mean(data, axis=1)

    # Log thông tin cơ bản
    print(f"File gốc: {os.path.basename(file_path)}")
    print(f"Tần số chuẩn: {E2_FREQ} Hz (E2)")

    # Tạo hai biến thể tăng và giảm cents
    high_cents = random.randint(5, 25)  # Tăng ngẫu nhiên từ 5 đến 25 cents
    low_cents = random.randint(5, 25)   # Giảm ngẫu nhiên từ 5 đến 25 cents

    # Tính toán hệ số tần số (factor) cho việc điều chỉnh pitch
    high_factor = cents_to_factor(high_cents)
    low_factor = cents_to_factor(-low_cents)

    # Điều chỉnh tần số và tạo dữ liệu mới
    high_data = adjust_pitch(data, high_factor)
    low_data = adjust_pitch(data, low_factor)

    # Ghi file đã điều chỉnh
    base_name = os.path.splitext(os.path.basename(file_path))[0]
    high_file_path = os.path.join(OUTPUT_FOLDER, f"{base_name}_high_{high_cents}cents.wav")
    low_file_path = os.path.join(OUTPUT_FOLDER, f"{base_name}_low_{low_cents}cents.wav")
    sf.write(high_file_path, high_data, sample_rate)
    sf.write(low_file_path, low_data, sample_rate)

    print(f"Đã tạo file: {high_file_path} (+{high_cents} cents)")
    print(f"Đã tạo file: {low_file_path} (-{low_cents} cents)")

# Chạy chương trình
if not os.path.exists(OUTPUT_FOLDER):
    os.makedirs(OUTPUT_FOLDER)

for file_name in os.listdir(INPUT_FOLDER):
    if file_name.endswith(".wav"):
        create_test_files(os.path.join(INPUT_FOLDER, file_name))
```
